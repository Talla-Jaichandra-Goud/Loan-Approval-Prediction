```python
# Feature 1 — cibil score category
bins = [300, 550, 650, 750, 900]
labels = ['poor', 'average', 'good', 'excellent']
X_train['cibil_category'] = pd.cut(X_train['cibil_score'], bins=bins, labels=labels)
X_test['cibil_category'] = pd.cut(X_test['cibil_score'], bins=bins, labels=labels)

# Feature 2 — total asset value
X_train['total_asset_value'] = X_train['residential_assets_value'] + X_train['commercial_assets_value'] + X_train['luxury_assets_value'] + X_train['bank_asset_value']
X_test['total_asset_value'] = X_test['residential_assets_value'] + X_test['commercial_assets_value'] + X_test['luxury_assets_value'] + X_test['bank_asset_value']

# Feature 3 — loan to income ratio
X_train['loan_to_income_ratio'] = X_train['loan_amount'] / X_train['income_annum']
X_test['loan_to_income_ratio'] = X_test['loan_amount'] / X_test['income_annum']

# Feature 4 — income category
bins = [0, 3000000, 6000000, 9900000]
labels = ['low', 'medium', 'high']
X_train['income_category'] = pd.cut(X_train['income_annum'], bins=bins, labels=labels)
X_test['income_category'] = pd.cut(X_test['income_annum'], bins=bins, labels=labels)

# Feature 5 — loan amount category
bins = [0, 10000000, 25000000, 39500000]
labels = ['low', 'medium', 'high']
X_train['loan_amount_category'] = pd.cut(X_train['loan_amount'], bins=bins, labels=labels)
X_test['loan_amount_category'] = pd.cut(X_test['loan_amount'], bins=bins, labels=labels)

print("New shape:", X_train.shape)
print(X_train[['cibil_category', 'total_asset_value', 'loan_to_income_ratio', 'income_category', 'loan_amount_category']].head())
```

FEATURE SELECTION

```python
# observing best numerical features
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import OrdinalEncoder

# Temporary encode categorical columns for Random Forest
X_temp = X_train.copy()
cat_cols_temp = X_temp.select_dtypes(include=['object','category']).columns.tolist()
X_temp[cat_cols_temp] = X_temp[cat_cols_temp].astype(str)

temp_enc = OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1)
X_temp[cat_cols_temp] = temp_enc.fit_transform(X_temp[cat_cols_temp])

# Train Random Forest
rf = RandomForestClassifier(n_estimators=200, random_state=42, n_jobs=-1)
rf.fit(X_temp, y_train)

# Feature importance
importance_df = pd.DataFrame({
    'feature': X_temp.columns,
    'importance': rf.feature_importances_
}).sort_values('importance', ascending=False)

print(importance_df)
```

```python
#observing best categorical features
from sklearn.feature_selection import chi2, SelectKBest
from sklearn.preprocessing import OrdinalEncoder

X_temp_cat = X_train.select_dtypes(include=['object','category']).copy()
X_temp_cat = X_temp_cat.astype(str)

enc = OrdinalEncoder(handle_unknown='use_encoded_value', unknown_value=-1)
X_temp_encoded = enc.fit_transform(X_temp_cat)

selector = SelectKBest(chi2, k='all')
selector.fit(X_temp_encoded, y_train)

chi2_scores = pd.DataFrame({
    'feature': X_temp_cat.columns,
    'chi2_score': selector.scores_
}).sort_values('chi2_score', ascending=False)

print(chi2_scores)
```

```python
#based on correlation heatmap,  RF importance checked, Chi-Square checked  we decided to drop. Even though we have many useless features we decided to drop few features only because we removed most useless features if the final accuracy is not good then we can decide to drop those remaining unimportant features

cols_to_drop = ['income_annum', 'luxury_assets_value', 'education',
                'self_employed', 'income_category', 'loan_amount_category']

X_train = X_train.drop(columns=cols_to_drop)
X_test = X_test.drop(columns=cols_to_drop)

print("New shape:", X_train.shape)
print(X_train.columns.tolist())
```

```python

from sklearn.preprocessing import StandardScaler, RobustScaler

standard_cols = ['loan_amount', 'cibil_score', 'total_asset_value',
                 'no_of_dependents', 'loan_term', 'cibil_category',
                 'loan_to_income_ratio']

robust_cols = ['residential_assets_value', 'commercial_assets_value',
               'bank_asset_value']

scaler_std = StandardScaler()
scaler_robust = RobustScaler()

X_train[standard_cols] = scaler_std.fit_transform(X_train[standard_cols])
X_test[standard_cols] = scaler_std.transform(X_test[standard_cols])

X_train[robust_cols] = scaler_robust.fit_transform(X_train[robust_cols])
X_test[robust_cols] = scaler_robust.transform(X_test[robust_cols])

print(X_train.head())
```

Model Building

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import f1_score

models = {
    'Logistic Regression': LogisticRegression(class_weight='balanced', max_iter=1000),
    'Decision Tree'      : DecisionTreeClassifier(class_weight='balanced', random_state=42),
    'Random Forest'      : RandomForestClassifier(class_weight='balanced', random_state=42),
    'XGBoost'            : XGBClassifier(random_state=42),
    'KNN'                : KNeighborsClassifier()
}

for name, model in models.items():
    model.fit(X_train, y_train)

    train_pred = model.predict(X_train)
    test_pred  = model.predict(X_test)

    train_f1 = f1_score(y_train, train_pred, average='weighted')
    test_f1  = f1_score(y_test,  test_pred,  average='weighted')
    gap      = train_f1 - test_f1

    print(f"{name}")
    print(f"  Train F1 : {train_f1:.4f}")
    print(f"  Test F1  : {test_f1:.4f}")
    print(f"  Gap      : {gap:.4f}")
    print()
```

```python
#selected models
from sklearn.model_selection import cross_val_score

shortlisted = {
    'Random Forest': models['Random Forest'],
    'XGBoost'      : models['XGBoost']
}

for name, model in shortlisted.items():
    cv_scores = cross_val_score(
        model, X_train, y_train,
        cv=5, scoring='f1_weighted')
    print(f"{name}")
    print(f"  CV F1 scores: {cv_scores}")
    print(f"  Mean CV F1 : {cv_scores.mean():.4f}")
    print()
```

```python
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score

best_model = models['Random Forest']
test_pred = best_model.predict(X_test)
test_prob = best_model.predict_proba(X_test)[:,1]

accuracy = accuracy_score(y_test, test_pred)
f1 = f1_score(y_test, test_pred, average='weighted')
roc_auc = roc_auc_score(y_test, test_prob)

print(f"Accuracy : {accuracy:.4f}")
print(f"F1 Score : {f1:.4f}")
print(f"ROC-AUC  : {roc_auc:.4f}")
```

HYPERPARAMETER TUNING

```python
from sklearn.model_selection import RandomizedSearchCV

param_dist = {
    'n_estimators'     : [100, 200, 300, 500],
    'max_depth'        : [None, 10, 20, 30],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf' : [1, 2, 4],
    'max_features'     : ['sqrt', 'log2']
}

search = RandomizedSearchCV(
    RandomForestClassifier(class_weight='balanced', random_state=42),
    param_distributions=param_dist,
    n_iter=50,
    cv=5,
    scoring='f1_weighted',
    random_state=42,
    n_jobs=-1
)

search.fit(X_train, y_train)

print("Best params:", search.best_params_)
print("Best CV F1 :", search.best_score_)
```

```python
tuned_model = search.best_estimator_

train_pred_tuned = tuned_model.predict(X_train)
test_pred_tuned  = tuned_model.predict(X_test)

train_f1 = f1_score(y_train, train_pred_tuned, average='weighted')
test_f1  = f1_score(y_test,  test_pred_tuned,  average='weighted')

print("Train F1 :", round(train_f1, 4))
print("Test F1  :", round(test_f1, 4))
print("Gap      :", round(train_f1 - test_f1, 4))
```

```python
from sklearn.metrics import RocCurveDisplay

RocCurveDisplay.from_estimator(
    tuned_model, X_test, y_test)
plt.title('ROC Curve')
plt.show()
```
